As we can see the prices are in paisa not in rupees we'll covert this in rupees


```python
# Look at a few price values
df[['mrp', 'discountedSellingPrice']].head(10)
```

```python
# Convert paise to rupees
df['mrp_rupees'] = df['mrp'] / 100
df['selling_price_rupees'] = df['discountedSellingPrice'] / 100
```

```python
df[['mrp', 'mrp_rupees',
    'discountedSellingPrice', 'selling_price_rupees']].head()
```

```python
df.shape
```

```python
df.drop(columns=["mrp","discountedSellingPrice"] ,inplace=True)
```
